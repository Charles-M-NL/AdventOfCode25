## Advent Of Code 2025
### Day 2: Gift Shop

```python
import regex as re
```

```python
# Part 1

# Open input
with open("Day2.txt", "r") as input:
    input = input.read().replace("\n", "").split(sep=",")

# Initialize counter
invalid_sum = 0

# Check IDs
for id_range in input:
    ids = range(int(id_range.split("-")[0]), int(id_range.split("-")[1]) + 1) # Create lists of ids in the range
    for id in ids:
        if re.fullmatch(r"(\d+)\1", str(id)): # Check double numbers
                invalid_sum += id # Add double number to counter
                    
# Print solution
print(f"The sum of all the invalid ID's is {invalid_sum}.")      
```
