## Advent Of Code 2025
### Day 4: Printing Department

```python
# Part 1

# Open input
with open("Day4.txt", "r") as input:
    input = input.read().splitlines()

dist = 1

forklift_count = 0

# Add boundaries to solve future indexing issues
for i in range(len(input)):
    input[i] = "x"*dist + input[i] + "x"*dist
input = ["x"*len(input[0])]*dist + input + ["x"*len(input[0])]*dist

# For each @, find the adjacent tiles and check amount of @
for row in range(1, len(input) - 1):
    for col in range(1, len(input[row]) - 1):
        if input[row][col] == "@":
            adjacent = ""
            for i in range(-dist, dist + 1):
                adjacent += input[row + i][col - dist: col + dist + 1]
            if adjacent.count("@") < 5:
                forklift_count += 1

# Print solution               
print(f"The amount of rolls of paper that can be accessed by a forklift is {forklift_count}.")
```

```python
# Part 2

# Open input
with open("Day4.txt", "r") as input:
    input = input.read().splitlines()

dist = 1

# Add boundaries to solve future indexing issues. Also turn strings into lists for easy removal later on
for i in range(len(input)):
    input[i] = ["x"]*dist + list(input[i]) + ["x"]*dist
input = [["x"]*len(input[0])]*dist + input + [["x"]*len(input[0])]*dist

forklift_count = 0
first_cycle = "yes"
remove_paper = []

# For each @, find the adjacent tiles and check amount of @
while len(remove_paper) > 0 or first_cycle == "yes":
    remove_paper = []
    first_cycle = "no"
    for row in range(1, len(input) - 1):
        for col in range(1, len(input[row]) - 1):
            if input[row][col] == "@":
                adjacent = []
                for i in range(-dist, dist + 1):
                    adjacent += input[row + i][col - dist: col + dist + 1]
                if adjacent.count("@") < 5:
                    forklift_count += 1
                    remove_paper.append((row, col))

    # Remove the paper
    for coords in remove_paper:
        input[coords[0]][coords[1]] = "R"

# Print solution                    
print(f"The amount of rolls of paper that can be accessed by a forklift is {forklift_count}.")
```
