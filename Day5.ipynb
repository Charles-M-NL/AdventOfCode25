## Advent Of Code 2025
### Day 5: Cafetaria

```python
# Part 1

# Open input
with open("Day5.txt", "r") as input:
    input = input.read().split("\n\n")

# Initialize counter    
fresh_count = 0

# Check if ingredient falls in a fresh range
for ingredient_id in [ingredient.strip() for ingredient in input[1].split("\n")[:-1]]:
    for fresh_range in [ingredient.strip() for ingredient in input[0].split("\n")[:-1]]:
        if int(fresh_range.split("-")[0]) <= int(ingredient_id) <= int(fresh_range.split("-")[1]):
            fresh_count += 1
            break

# Print solution        
print(f"The amount of available ingredient IDs that are fresh is {fresh_count}.")
```

```python
# Part 2

# Open input
with open("Day5.txt", "r") as input:
    lines = input.readlines()
    
# Split ranges from ingredient ids
ranges = [[int(n) for n in line.split('-')] for line in lines[:lines.index('\n')]]

# For each range, check if there is a range where B>=C and A<=D (considering ranges A-B and C-D) 
for i, range0 in enumerate(ranges):
    for j, range1 in enumerate(ranges[i + 1:], start=i + 1):
        if range0[1] >= range1[0] and range0[0] <= range1[1]:
            # Set one of the ranges to [1 ,0]
            ranges[i] = [1, 0]
            # Set the other range to account for both original ranges
            ranges[j] = [min(range0[0], range1[0]), max(range0[1], range1[1])]
            break

# We keep the absorbed ranges to not break the indexing. We set them to [1, 0] to make them count for -1, to counteract the + len(ranges) we need at the end to account for the inclusiveness of the ranges

# Print solution
print(sum([range_[1] - range_[0] for range_ in ranges]) + len(ranges))
```
