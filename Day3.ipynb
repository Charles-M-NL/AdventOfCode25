## Advent Of Code 2025
### Day 3: Lobby

```python
# Part 1

# Open input
with open("Day3.txt", "r") as input:
    input = input.read().splitlines()

# Initialize counter
total_joltage = 0    

# Calculate max joltage per bank and add to counter    
for bank in input:
    
    i = max([int(battery) for battery in bank[:-1]])
    j = max([int(battery) for battery in bank[bank.index(str(i)) + 1:]])
    
    total_joltage += int(str(i) + str(j))
    
# Print solution           
print(f"The total output joltage is {total_joltage}.")
```

```python
# Part 2

# Open input
with open("Day3.txt", "r") as input:
    input = input.read().splitlines()

# Initialize counter
total_joltage = 0      

# Calculate max joltage per bank and add to counter    
for bank in input:
    max_joltage = ""
    i = 0
    joltage_length = 12 # Parameter of the length so it can be solved for any amount of batteries. Joltage_length = 2 solves part 1
    # First 11 times: find max within the range from the previously chosen number to however far you can go while leaving enough numbers for the rest of the loop
    for j in range(joltage_length - 1):   
        if i < 99:
            k = max([int(battery) for battery in bank[i: ((joltage_length - 1 - j)*-1)]])
            max_joltage += str(k)
            i = i + 1 + bank[i:].index(str(k))
        else:
            max_joltage += bank[-1]
    # One last time with whatever is left of the string
    k = max([int(battery) for battery in bank[i:]])
    max_joltage += str(k)
                
    total_joltage += int(max_joltage)
    
# Print solution           
print(f"The total output joltage is {total_joltage}.") 
```
