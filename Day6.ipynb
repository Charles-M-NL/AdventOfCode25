## Advent Of Code 2025
### Day 6: Trash Compactor

```python
import numpy as np
import regex as re
```

```python
# Part 1

# Open input
with open("Day6.txt", "r") as input:
    input = input.read().splitlines()

# Intitialize counter
grand_total = 0

# Split the input lines on any whitespace
input = [re.split(r"\s+", row) for row in input]

# Take the numbers of each column and perform the operation
for i in range(len(input[0])):
    nums = [int(row[i]) for row in input[:-1]]
    if input[-1][i] == "+":
        grand_total += np.sum(nums)
    elif input[-1][i] == "*":
        grand_total += np.prod(nums)

# Print solution
print(f"The grand total found by adding together all of the answers is {grand_total}.")
```

```python
# Part 2

# Open input
with open("Day6.txt", "r") as input:
    input = input.read().splitlines()

# Initialize counter and column list to fill with numbers read vertically
grand_total = 0
column_list = []

# Fill the column list
for col in range(len(input[0])):
    column = ""
    for row in range(len(input))[:-1]:
        column += input[row][col]
    column_list.append(column)

# Add a separator to later separate the numbers, after splitting them into groups
sep_nums = "-".join(column_list).split("    ")
nums = [re.sub(r"\s+", "", line).split("-") for line in sep_nums]

# Trim the edges of the groups. There are "" entries at the start and ends of most groups
trimmed_nums = []
for line in nums:
    trimmed_nums.append([int(num) for num in line if num != ""])

# Define the operator list
operator_list = re.split(r"\s+", input[-1])[:-1]

# Perform arithmetic
for i in range(len(trimmed_nums)):
    if operator_list[i] == "+":
        grand_total += np.sum([trimmed_nums[i]])
    elif operator_list[i] == "*":
        grand_total += np.prod([trimmed_nums[i]])

# Print solution
print(f"The grand total found by adding together all of the answers is {grand_total}.")
```
