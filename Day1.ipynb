## Advent Of Code 2025
### Day 1: Secret Entrance

```python
# Part 1

# Open input
with open("Day1.txt", "r") as input:
    input = input.read().splitlines()

# Define the dial list and starting position    
dial = list(range(0, 100))
dial_index = 50

# Start the counter
zero_counter = 0

# Turn the dial
for turn in input:
    if turn[0] == "L":
        dial_index = (dial_index - int(turn[1:])) % len(dial) 
    elif turn[0] == "R":
        dial_index = (dial_index + int(turn[1:])) % len(dial) 
    
    # Check if the dial is left pointing at 0
    if dial_index == 0:
        zero_counter += 1

# Print solution
print(f"The amount of times the dial was left pointing at 0 is {zero_counter}.")
```

```python
# Part 2

# Open input
with open("Day1.txt", "r") as input:
    input = input.read().splitlines()

# Define the dial list and starting position    
dial = list(range(0, 100))
dial_index_list = [50]

# Start the counter
zero_counter = 0

# Turn the dial
for turn in input:
    if turn[0] == "L":
        for i in range(int(turn[1:])):
            dial_index_list.append(dial_index_list[-1] - 1)
    elif turn[0] == "R":
        for i in range(int(turn[1:])):
            dial_index_list.append(dial_index_list[-1] + 1) 
    
# Check if the dial is left pointing at 0
for i in dial_index_list:
    if i % 100 == 0:
        zero_counter += 1

# Print solution
print(f"The amount of times the dial pointed at 0 is {zero_counter}.")
```
